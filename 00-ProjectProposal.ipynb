## Research Question

Is there a statistically significant correlation between the funding levels of public college districts in the United States and the median annual earnings of their graduates ten years post-graduation, while accounting for regional cost-of-living differences?


## Background and Prior Work

Public higher education plays a central role in shaping economic mobility and workforce outcomes in the United States. Public colleges and universities educate a substantial proportion of undergraduate students and are primarily funded through a combination of state appropriations, federal support, and tuition revenue. However, funding levels vary widely across institutions due to differences in state budgets, enrollment patterns, and regional economic priorities. According to the National Center for Education Statistics (NCES), public institutions receive markedly different levels of per-student revenue depending on state and local funding structures, raising concerns about unequal access to educational resources and long-term economic opportunity for students.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

Prior research and government reporting have examined the relationship between higher education and post-graduation earnings, generally finding that college attendance is associated with higher lifetime income. The U.S. Department of Education’s College Scorecard aggregates administrative earnings data to report median earnings of graduates several years after enrollment, enabling systematic comparisons across institutions.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) NCES reporting further documents how institutional revenues are allocated toward instruction, academic support, and student services, suggesting potential mechanisms through which funding levels may influence long-term student outcomes.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)

Several prior COGS 108 projects have investigated related questions about the economic value of higher education. For example, the project *Are Universities Worth the Opportunity Cost?* analyzed whether college attendance yields sufficient financial returns compared to entering the workforce directly, concluding that earnings outcomes vary substantially by institutional and contextual factors rather than education alone.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) This work highlights the importance of examining institutional characteristics—such as funding levels—rather than treating higher education as a uniform experience.

Despite the availability of these datasets, existing work faces notable limitations. A major threat to validity is omitted variable bias, as students from higher socioeconomic backgrounds may be more likely to attend better-funded institutions, meaning post-graduation earnings may reflect pre-existing advantages rather than institutional investment alone. Additionally, regional labor market conditions and cost-of-living differences independently influence earnings outcomes. Our project builds on prior work by focusing explicitly on institutional funding while incorporating regional socioeconomic controls, allowing us to interpret observed relationships as correlational rather than causal.

## Hypothesis

We predict a positive correlation between institutional funding levels and long-term median earnings of graduates. Higher funding may allow institutions to maintain lower student-to-teacher ratios and invest more heavily in instructional quality and vocational resources, thereby enhancing students’ human capital and earning potential. Given regional disparities in funding, we expect higher-funded public institutions to be better equipped to provide resources that translate into improved labor market outcomes.


## Data

**Variables:**
- Institution-level funding or revenue (proxy for per-student expenditure)
- Median earnings of graduates ten years after enrollment
- Institutional location (state or region)
- Enrollment size and degree completions
- Tuition and cost of attendance
- Regional socioeconomic indicators (e.g., median household income)

**Observations:**
At least 1,000 public colleges and universities across the United States to ensure sufficient statistical power and geographic diversity.

**Collection:**
Data will be aggregated from NCES administrative datasets (College Scorecard and IPEDS) and supplemented with socioeconomic data from the U.S. Census Bureau’s American Community Survey.

**Storage:**
Data will be stored in a tidy CSV format with one row per institution and standardized identifiers to enable reproducible merging and analysis.


## Ethics & Privacy

All ethical checklist items have been considered. This project relies exclusively on publicly available, aggregated institutional data with no personally identifiable information. Potential downstream harms, such as the misuse of correlational findings to justify funding reductions, will be explicitly addressed by clearly communicating limitations and avoiding causal claims.
