## Research Question

Is there a statistically significant correlation between the funding levels of public college districts in the U.S. and the median annual earnings of their graduates ten years post-graduation?


## Background and Prior Work

The funding of public higher education remains a central topic in debates regarding economic advancement, workforce outcomes, and the promotion of financial mobility in the United States.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Public colleges and universities educate a significant proportion of the nation’s working population, as they receive financial backing from an integration of state, tuition, and federal funding.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) However, this degree of funding varies across public college and university districts as a result of differences in state budgets, enrollment rates, and regional economic priorities.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) These considerable variations have prompted our discussion of whether greater financial investment in higher education systems leads to stronger long-term economic outcomes for their graduates.

From NCES, we extracted five datasets: median earnings of graduates, institutional funding and revenue data, tuition and cost of attendance, enrollment and completions, and institutional characteristics such as location and size.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Together, these datasets allow us to examine the relationship between funding levels and graduate earning while accounting for institutional scale and geographic differences.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) To further contextualize earnings outcomes, we incorporate cost-of-living information from the U.S. Census Bureau, enabling a more accurate comparison across regions.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)

A prior COGS 108 project (“Are Universities Worth the Opportunity Cost?”) explored education outcomes and economic tradeoffs, providing a useful reference for methods and framing around returns to higher education using institutional-level data.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)

## Hypothesis


We predict a positive correlation between the level of school funding and long-term median earnings. Our thinking revolves around the idea that higher funding allows for lower student-to-teacher ratios and better access to vocational resources, which theoretically enhances the human capital and earning potential of students. When taking into account the different regional levels of funding for various higher education institutions mentioned in Background, higher-funded institutions are generally more capable of providing resources that can equip students with skills and experiences that are valuable in the labor market.


## Data

### Ideal Dataset

The ideal dataset would include institution-level funding or revenue measures (as a proxy for per-student expenditure), linked to long-term earnings outcomes for graduates ten years post-graduation, with consistent institutional identifiers for merging across sources.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) Additional variables would include institutional location (state/region), enrollment and degree completions, tuition and cost of attendance, and demographic and socioeconomic context variables (e.g., regional median household income) to serve as controls.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) We would aim for at least 1,000 public colleges and universities across the United States to ensure sufficient statistical power and geographic diversity.

Collection would rely on federal education databases, including NCES administrative records (College Scorecard and IPEDS) and socioeconomic data from the U.S. Census Bureau’s American Community Survey (ACS).<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6)<a name="cite_ref-7"></a>[<sup>7</sup>](#cite_note-7)<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) The data would be stored in a tidy CSV format with one row per institution and standardized identifiers (e.g., IPEDS UnitID and state) to allow merging across datasets.

### Potential Real Datasets (Sources)

**1) College Scorecard (U.S. Department of Education)**<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6)
- **Where**: Public bulk download/API at https://collegescorecard.ed.gov/data
- **Why it’s useful**: Provides outcome measures including median earnings (including measures around ~10 years after entry for many institutions) and institutional identifiers (including IPEDS UnitID) that can be merged with IPEDS.
- **Key variables**: Earnings outcomes (median earnings), institution identifiers (UNITID/opeid), and other institutional metrics.

**2) IPEDS (NCES) – Finance / Institutional Characteristics / Enrollment & Completions**<a name="cite_ref-7"></a>[<sup>7</sup>](#cite_note-7)<a name="cite_ref-8"></a>[<sup>8</sup>](#cite_note-8)
- **Where**: IPEDS Data Center and survey components: https://nces.ed.gov/ipeds/
- **Why it’s useful**: Includes institutional funding/revenue variables (e.g., total revenue, appropriations, expenditures), plus institutional characteristics and scale measures (enrollment/completions) needed for controls.
- **Key variables**: Total revenue, instructional expenditures, state/federal appropriations, tuition/cost of attendance, enrollment, completions, location, sector.

**3) U.S. Census Bureau – American Community Survey (ACS) 5-year estimates**<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)
- **Where**: https://data.census.gov/
- **Why it’s useful**: Provides regional socioeconomic context variables as proxies for cost-of-living differences; we can use median household income as a control when comparing across regions.
- **Key variables**: Median household income (ACS Table B19013), plus other demographic/economic indicators at geographic levels (state/county/metro as needed).

## Ethics 

### A. Data Collection
 - [X] **A.1 Informed consent**: Our project used publicly available datasets collected by federal agencies. Since we had no direct interaction with the human subjects, informed consent was handled by the original institutions under their legal and ethical frameworks.
 - [X] **A.2 Collection bias**: There may be collection bias due to how different districts report funding and how earnings are aggregated geographically. Additionally, missing or incomplete records may be systematically different from districts with complete data. We will address this concern by documenting missing data and interpreting results cautiously.
 - [X] **A.3 Limit PII exposure**: No PII is used in this project. All data are aggregated at the school district/geographic level which prevents identification of individuals.
 - [X] **A.4 Downstream bias mitigation**: When available, variables such as race/ethnicity and gender distributions will be included to examine how outcomes may differ across these groups.

### B. Data Storage
 - [X] **B.1 Data security**: We are relying on publicly available data sources, plan on keeping libraries up to date, and are planning to avoid downloading unnecessary or sensitive data.
 - [X] **B.2 Right to be forgotten**: Not applicable, we are relying on publicly available information which does not include personally identifiable information.
 - [X] **B.3 Data retention plan**: Not applicable, we are relying on publicly available information which does not include personally identifiable information.

### C. Analysis
 - [X] **C.1 Missing perspectives**: The datasets we are using won’t be able to cover other contextual factors such as job opportunities locally, higher education accessibility, or other community support systems. We will acknowledge these limitations in our discussion.
 - [X] **C.2 Dataset bias**: We will take steps to address these biases by explicitly stating observed correlations may reflect broader socioeconomic forces rather than a direct casual effect of school funding.
 - [X] **C.3 Honest representation**: Our visualizations, summary statistics, and reports will be designed to honestly represent the underlying data.
 - [X] **C.4 Privacy in analysis**: Have we ensured that data with PII are not used or displayed unless necessary for the analysis? No PII is used or displayed.
 - [X] **C.5 Auditability**: The process of generating the analysis will be well documented and reproducible if we discover issues in the future.

### D. Modeling
 - [X] **D.1 Proxy discrimination**: Yes, we are taking measures to reduce proxy discrimination. Our model uses district-level funding rather than individual characteristics and avoids direct use of protected attributes like race or gender. We also use regional cost-of-living differences to limit geographic variables from acting as proxies for differing wealth demographics.
 - [X] **D.2 Fairness across groups**: To assess fairness across groups, we will evaluate whether model errors differ systematically across types of institutions (such as public versus private colleges, institution size, or geographic region), ensuring that the model does not consistently over-or under-estimate earnings for particular groups of schools.
 - [X] **D.3 Metric selection**: We considered how optimizing for our primary performance metrics (such as prediction error) could oversimplify outcomes, so we also examine additional metrics like error distribution across institutions and uncertainty estimates to ensure results are accurate, balanced, and not misleading.
 - [X] **D.4 Explainability**: Yes, the model’s outputs can be explained in understandable terms by describing how institutional funding and other included factors contributed to the predicted earnings, while clearly communicating the model’s assumptions, limitations, and uncertainty rather than presenting results as definitive decisions.
 - [X] **D.5 Communicate limitations**: We communicate the model’s limitations, potential biases, and uncertainties in plain language alongside the results, emphasizing that findings are correlational, context-dependent, and should not be interpreted as definitive or causal.

### E. Deployment
 - [X] **E.1 Monitoring and evaluation**: After deployment, we will regularly check the model’s accuracy with new data, review a sample of predictions, require human oversight for high-stakes uses, and watch for data changes or unintended negative impacts over time.
 - [X] **E.2 Redress**: If users are negatively affected by the results, we would carefully review the situation, assess whether the data or modeling choices contributed, and make appropriate updates to the analysis to reduce the chance of similar harm in the future.
 - [X] **E.3 Roll back**: Since there are no plans for large-scale deployment beyond use within an academic setting, our analyses and conclusions can be readily rolled back if needed.
 - [X] **E.4 Unintended use**: We can see that our findings could be misused to oversimplify debates about school funding. To address this risk, we will emphasize caution regarding our conclusions and the many different factors that may also be influencing long-term earning outcomes beyond what we analyzed.
