## Research Question

Is there a statistically significant correlation between the funding levels of public college districts in the United States and the median annual earnings of their graduates ten years post-graduation?


## Background and Prior Work

The funding of public higher education remains a central topic in debates regarding economic advancement, workforce outcomes, and the promotion of financial mobility in the United States. Public colleges and universities educate a significant proportion of the nation’s working population and receive financial backing through a combination of state appropriations, federal support, and tuition revenue. However, funding levels vary substantially across public college and university districts due to differences in state budgets, enrollment trends, and regional economic priorities. These variations raise important questions about whether greater financial investment in public higher education systems leads to stronger long-term economic outcomes for their graduates.

Previous work and publicly available administrative datasets from the National Center for Education Statistics (NCES) provide a foundation for examining this relationship. From NCES, we draw on datasets covering median earnings of graduates, institutional funding and revenue, tuition and cost of attendance, enrollment and degree completions, and institutional characteristics such as location and size. Together, these datasets allow us to examine the association between institutional funding levels and graduate earnings while accounting for differences in institutional scale and geographic context. To further contextualize earnings outcomes, we incorporate regional cost-of-living proxies using socioeconomic data from the U.S. Census Bureau, allowing for more meaningful comparisons across regions.

This project builds on prior analyses that examine the economic returns to higher education and the opportunity costs associated with attending college. For example, a previous COGS 108 project investigated whether universities are worth their opportunity cost, providing a useful methodological reference for analyzing education-related economic outcomes using institutional-level data.

## Hypothesis

We predict a positive correlation between institutional funding levels and long-term median earnings of graduates. Higher funding may allow institutions to maintain lower student-to-teacher ratios, invest in instructional quality, and provide stronger vocational and career preparation resources. These investments may enhance students’ human capital and improve their earning potential. Given regional disparities in funding, we expect higher-funded public institutions to be better equipped to offer resources that translate into improved labor market outcomes for graduates.


## Data

**Variables:**
- Institution-level funding or revenue (used as a proxy for per-student expenditure)
- Median earnings of college graduates several years after completion
- Institutional location (state or region)
- Enrollment size and degree completions
- Tuition and cost of attendance
- Regional socioeconomic context variables (e.g., median household income)

**Observations:**
At least 1,000 public colleges and universities across the United States to ensure sufficient statistical power and geographic diversity.

**Collection:**
Data will be aggregated from federal education databases, including the College Scorecard and IPEDS datasets maintained by NCES, along with socioeconomic data from the U.S. Census Bureau’s American Community Survey. Earnings data are derived from linked administrative sources, while funding and institutional characteristics are reported directly by institutions.

**Storage:**
The final dataset will be stored in a tidy CSV format with one row per institution and standardized identifiers (such as institution ID and state) to facilitate merging across datasets.
