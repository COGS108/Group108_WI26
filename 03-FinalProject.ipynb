## Abstract

Public college funding is often discussed as a driver of educational quality and economic mobility, but it remains unclear whether higher institutional spending is associated with stronger long-term earnings outcomes for graduates. In this project, we investigate whether public colleges with higher instructional funding per student tend to have higher median graduate earnings approximately ten years after enrollment.

To address this question, we combined institution-level financial and enrollment data from the Integrated Postsecondary Education Data System (IPEDS), graduate earnings and geographic identifiers from the U.S. Department of Education College Scorecard, and state-level median household income from the American Community Survey (ACS). These datasets were merged using institutional identifiers and state abbreviations to create a final analytic dataset of 631 public institutions. We conducted exploratory data analysis to examine distributions, identify outliers, and assess the relationship between funding and earnings. Because both funding and earnings were right-skewed, we used log-transformed variables in subsequent analyses.

Our exploratory analyses revealed a positive association between instructional funding per student and graduate earnings. The relationship became clearer after log transformation, with the correlation between log funding and log earnings stronger than the correlation in the original scale. Additional analyses suggested that institutional enrollment was only weakly related to both funding and earnings, while geographic and regional factors contributed modestly to earnings differences across institutions.

Overall, our findings suggest that better-funded public colleges tend to be associated with higher long-term graduate earnings, although the relationship is not purely causal and may also reflect regional labor market conditions and other institutional differences. These results support the importance of institutional resources while also motivating more detailed regression modeling and discussion of possible confounding factors.

## Research Question

Is there a statistically significant association between institutional funding levels at U.S. public colleges and the median annual earnings of graduates approximately ten years after enrollment?

Funding will be measured using institution-level financial variables such as total revenue, instructional expenditures, and state and federal appropriations obtained from IPEDS/NCES. Graduate outcomes will be measured using the College Scorecard median earnings metric at approximately ten years after completion.

To account for potential confounding factors, we will include controls for institutional scale (enrollment and completions), geographic variation (state/region), and regional cost-of-living proxies (ACS median household income). This is a statistical inference project using correlation and regression analyses.

## Background and Prior Work

The funding of public higher education remains a central topic in debates regarding economic advancement, workforce outcomes, and the promotion of financial mobility in the United States. Public colleges and universities educate a significant proportion of the nation’s working population, as they receive financial backing from an integration of state support, tuition revenue, and federal funding.<a id="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) However, this degree of funding varies across public college and university districts as a result of differences in state budgets, enrollment rates, and regional economic priorities.<a id="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) These variations motivate our question of whether greater financial investment in higher education systems is associated with stronger long-term economic outcomes for graduates.

To study this relationship, we plan to use publicly available administrative datasets from the National Center for Education Statistics (NCES) and related federal education data systems. From NCES, we extracted or identified datasets that capture: (1) median earnings of graduates<a id="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6), (2) institutional funding and revenue, (3) tuition and cost of attendance, (4) enrollment and completions, and (5) institutional characteristics such as location and size.<a id="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Together, these datasets allow us to examine how funding levels relate to graduate earnings while accounting for institutional scale and geographic differences.

A key concern in comparing earnings across institutions is that similar salaries can imply different standards of living depending on local costs and regional economic conditions. To further contextualize earnings outcomes, we incorporate cost-of-living information using socioeconomic data from the U.S. Census Bureau (American Community Survey), enabling a more accurate comparison of earnings across regions by using a regional proxy such as median household income.<a id="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)

This project also connects to prior work that evaluates the economic value of higher education, including student earnings outcomes and the opportunity costs of attending college. A previous COGS 108 project (“Are Universities Worth the Opportunity Cost?”) provides a relevant methodological reference for analyzing education-related economic outcomes using institutional-level data, and it motivates our emphasis on measurable labor market outcomes in a rigorous, data-driven framework.<a id="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5)

## Hypothesis

We predict a positive association between institutional funding per student and median earnings ten years after entry. Higher funding may allow institutions to provide better academic and career resources that improve students’ long-term earning potential. Because earnings vary substantially by major, we will control for the types of degrees offered at each institution. Specifically, we will include measures of degree composition (e.g., the proportion of graduates in higher-paying fields such as engineering, computer science, health, or business) using data from the College Scorecard. This helps ensure that any observed relationship between funding and earnings is not simply due to certain schools offering more high-paying majors. We expect funding to remain positively associated with earnings even after accounting for institutional size, geography, and degree mix.

## Results

### Exploratory Data Analysis

Exploratory data analysis (EDA) was performed to examine distributions, identify outliers, and understand relationships between funding, earnings, enrollment, and geographic variables.

The full EDA process is shown in `02-EDACheckpoint.ipynb`. In this report we summarize the key findings relevant to our research question.

All analyses below use the final analytic dataset (`df`) containing 631 institutions.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
sns.histplot(df["funding_per_student_2012"], bins=40)
plt.title("Funding per Student Distribution")
plt.show()


plt.figure(figsize=(6,4))
sns.histplot(df["MD_EARN_WNE_P10"], bins=40)
plt.title("Earnings Distribution")
plt.show()
```

Both funding per student and graduate earnings show right-skewed distributions with several extreme values. Because of this skewness, log transformations were used in later analyses.

```python
import numpy as np

df["log_funding"] = np.log10(df["funding_per_student_2012"])
df["log_earnings"] = np.log10(df["MD_EARN_WNE_P10"])

plt.figure(figsize=(6,4))
sns.histplot(df["log_funding"], bins=40)
plt.title("Log Funding")
plt.show()

plt.figure(figsize=(6,4))
sns.histplot(df["log_earnings"], bins=40)
plt.title("Log Earnings")
plt.show()
```

Log transformation reduces skewness and produces more symmetric distributions, making correlation and regression analysis more appropriate.

```python
plt.figure(figsize=(6,5))

sns.scatterplot(
    x="log_funding",
    y="log_earnings",
    data=df,
    alpha=0.6
)

plt.title("Log Funding vs Log Earnings")
plt.show()
```

There is a positive association between funding per student and graduate earnings. The relationship becomes clearer after log transformation.

```python
plt.figure(figsize=(6,5))

sns.heatmap(
    df[[
        "funding_per_student_2012",
        "MD_EARN_WNE_P10",
        "enrollment_2012"
    ]].corr(),
    annot=True,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix")
plt.show()
```

Funding per student is positively correlated with graduate earnings, while enrollment shows weaker correlations with both funding and earnings. This suggests that institutional size alone does not explain the observed association. The correlation between funding per student and graduate earnings is moderate in the raw scale and becomes stronger after log transformation, further supporting the use of log-transformed regression models.

### Regression Analysis: Funding and Graduate Earnings

To formally evaluate the relationship between institutional funding and graduate earnings, we estimate linear regression models using log-transformed funding and earnings variables. Log transformation is appropriate because both variables are right-skewed and the EDA suggested a more linear association in log space.

We begin with a simple bivariate model and then add controls for enrollment, geographic region, and state median household income.

```python
import statsmodels.formula.api as smf
```

```python
model1 = smf.ols(
    formula="log_earnings ~ log_funding",
    data=df
).fit()

model1.summary()
```

Model 1 estimates the bivariate relationship between log funding per student and log graduate earnings. A positive coefficient on `log_funding` would indicate that institutions with higher funding per student tend to have higher median earnings outcomes.

```python
model2 = smf.ols(
    formula="log_earnings ~ log_funding + enrollment_2012 + median_household_income + C(REGION)",
    data=df
).fit()

model2.summary()
```

Model 2 adds controls for institutional enrollment, regional cost-of-living proxy (`median_household_income`), and Census region fixed effects. This model provides a more realistic estimate of the association between institutional funding and graduate earnings after accounting for geographic and institutional differences.

```python
results_table = pd.DataFrame({
    "Model": ["Model 1", "Model 2"],
    "log_funding_coef": [model1.params["log_funding"], model2.params["log_funding"]],
    "log_funding_pval": [model1.pvalues["log_funding"], model2.pvalues["log_funding"]],
    "R_squared": [model1.rsquared, model2.rsquared]
})

results_table
```

Across both models, the coefficient on `log_funding` indicates whether funding remains positively associated with graduate earnings after controls are added. The p-value indicates whether this association is statistically significant, while the R-squared value summarizes how much variation in earnings is explained by the model.

```python
plt.figure(figsize=(6,5))
sns.regplot(
    x="log_funding",
    y="log_earnings",
    data=df,
    scatter_kws={"alpha": 0.5}
)
plt.title("Regression Relationship Between Log Funding and Log Earnings")
plt.show()
```

The regression results show a strong positive association between institutional funding per student and median graduate earnings.

In Model 1, which includes only log funding as a predictor, the coefficient on `log_funding` is 0.302 (p < 0.001), indicating that institutions with higher funding per student tend to have higher earnings outcomes. The R² value of 0.44 suggests that funding alone explains about 44% of the variation in log earnings across institutions.

In Model 2, we add controls for enrollment, geographic region, and state median household income. The coefficient on `log_funding` remains positive and highly significant (β = 0.273, p < 0.001), showing that the relationship between funding and earnings persists even after accounting for institutional size and regional economic differences. The R² increases to 0.53, indicating that the additional control variables improve the explanatory power of the model.

Median household income is also positively associated with earnings, suggesting that regional economic conditions influence graduate outcomes. Enrollment has a small but significant positive coefficient, while regional indicators show systematic differences across geographic areas.

Overall, these results support our hypothesis that higher institutional funding per student is associated with higher long-term graduate earnings, although the relationship is not purely causal and likely reflects both institutional resources and regional labor market conditions.

### Secondary Analysis: Funding Quartiles and Earnings

To further evaluate the relationship between funding and earnings, we divide institutions into quartiles based on funding per student. This allows us to compare earnings outcomes across low-funded and high-funded institutions without assuming a strictly linear relationship.

If funding is related to earnings, we expect institutions in higher funding quartiles to have higher median graduate earnings.

```python
df["funding_quartile"] = pd.qcut(
    df["funding_per_student_2012"],
    4,
    labels=["Q1 Low", "Q2", "Q3", "Q4 High"]
)

df[["funding_per_student_2012", "funding_quartile"]].head()
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(
    x="funding_quartile",
    y="MD_EARN_WNE_P10",
    data=df,
    order=["Q1 Low", "Q2", "Q3", "Q4 High"]
)

plt.title("Graduate Earnings by Funding Quartile")
plt.ylabel("Median Earnings")
plt.xlabel("Funding Quartile")

plt.show()
```

The boxplot shows a clear increase in graduate earnings as funding per student increases. Institutions in the highest funding quartile have noticeably higher median earnings than those in the lowest quartile. 

This pattern supports the regression results and suggests that the positive association between funding and earnings is not driven by a small number of extreme institutions. Instead, the relationship appears consistently across funding levels.

```python
df.groupby("funding_quartile")["MD_EARN_WNE_P10"].mean()
```

Average earnings increase steadily from the lowest to the highest funding quartile, providing additional evidence that higher institutional spending is associated with higher graduate earnings.
